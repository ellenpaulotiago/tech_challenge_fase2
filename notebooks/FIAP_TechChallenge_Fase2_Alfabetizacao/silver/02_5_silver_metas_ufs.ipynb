## 1.3 Comparação dos Tipos de Dados

**Contexto**

Essa análise permite identificar alterações de esquema, inclusão ou remoção

**Objetivo**

Comparar os tipos de dados das bases de Metas por UF dos anos de 2023,

**Resultado esperado**

Obter um relatório consolidado contendo os tipos de dados de cada coluna e

````python
# Objetivo:
# Comparar os tipos de dados das
# bases de Metas por UF dos anos de
# 2023, 2024 e 2025.

# Justificativa:
# Colunas com o mesmo nome podem
# apresentar tipos diferentes entre
# os anos ou novas variáveis podem
# ter sido incorporadas pelo INEP.

# Ação:
# Consolida os tipos de dados das
# três bases em um único relatório
# e classifica a compatibilidade
# entre os esquemas.

relatorio_dtypes = pd.DataFrame({
    "2023": df_metas_ufs_2023.dtypes.astype(str),
    "2024": df_metas_ufs_2024.dtypes.astype(str),
    "2025": df_metas_ufs_2025.dtypes.astype(str)
})

relatorio_dtypes = relatorio_dtypes.replace("nan", pd.NA)


def classificar_status(linha):

    if (
        pd.notna(linha["2023"])
        and pd.isna(linha["2024"])
        and pd.isna(linha["2025"])
    ):
        return "Exclusiva de 2023"

    if (
        pd.isna(linha["2023"])
        and pd.notna(linha["2024"])
        and pd.notna(linha["2025"])
    ):
        return "Nova a partir de 2024"

    if (
        pd.isna(linha["2023"])
        and pd.isna(linha["2024"])
        and pd.notna(linha["2025"])
    ):
        return "Nova em 2025"

    tipos = linha.dropna()

    if len(tipos.unique()) == 1:
        return "Compatível"

    return "Divergente"


relatorio_dtypes["status"] = relatorio_dtypes.apply(
    classificar_status,
    axis=1
)

display(relatorio_dtypes)
````

## 1.7 Validação da granularidade estadual

As bases de metas por UF podem conter um registro agregado do Brasil sem código de UF.

Esse registro é válido como indicador nacional, mas não pertence à granularidade estadual. Para preservar a consistência da camada Silver:

- registros com `CD_UF` preenchido seguem para a base estadual;
- o agregado Brasil é preservado separadamente;
- registros sem código de UF que não representem o Brasil são enviados para rejeitados.

````python
# Objetivo:
# Separar os registros estaduais,
# o agregado Brasil e os registros
# inválidos da entidade Metas por UF.

# Justificativa:
# A ausência de código de UF é
# legítima apenas para o agregado
# nacional. A base Silver estadual
# deve manter granularidade exclusiva
# por UF.

# Ação:
# Classifica os registros com base no
# preenchimento de CD_UF e na
# identificação territorial.

def mascara_valor_ausente(serie):

    return (
        serie.isna()
        | (
            serie
            .astype(str)
            .str.strip()
            .str.lower()
            .isin([
                "",
                "nan",
                "none",
                "<na>",
                "null"
            ])
        )
    )


def identificar_coluna_territorio(df):

    for coluna in [
        "NOME_UF",
        "NO_UF",
        "SIGLA_UF",
        "SG_UF"
    ]:
        if coluna in df.columns:
            return coluna

    return None


def separar_metas_ufs(
    df,
    ano
):

    df_trabalho = df.copy()

    coluna_codigo = (
        "CD_UF"
        if "CD_UF" in df_trabalho.columns
        else "CO_UF"
    )

    coluna_territorio = (
        identificar_coluna_territorio(
            df_trabalho
        )
    )

    codigo_ausente = (
        mascara_valor_ausente(
            df_trabalho[coluna_codigo]
        )
    )

    if coluna_territorio is not None:

        territorio_normalizado = (
            df_trabalho[
                coluna_territorio
            ]
            .astype(str)
            .str.strip()
            .str.upper()
        )

        eh_brasil = (
            territorio_normalizado
            .isin([
                "BRASIL",
                "BR",
                "TOTAL BRASIL"
            ])
        )

    else:
        eh_brasil = pd.Series(
            False,
            index=df_trabalho.index
        )

    mascara_brasil = (
        codigo_ausente
        & eh_brasil
    )

    mascara_rejeitados = (
        codigo_ausente
        & ~eh_brasil
    )

    df_validos = (
        df_trabalho[
            ~codigo_ausente
        ]
        .copy()
        .reset_index(drop=True)
    )

    df_brasil = (
        df_trabalho[
            mascara_brasil
        ]
        .copy()
        .reset_index(drop=True)
    )

    df_rejeitados = (
        df_trabalho[
            mascara_rejeitados
        ]
        .copy()
        .reset_index(drop=True)
    )

    if not df_rejeitados.empty:

        df_rejeitados[
            "motivo_rejeicao"
        ] = (
            "CODIGO_UF_NULO_"
            "SEM_IDENTIFICACAO_BRASIL"
        )

        df_rejeitados[
            "_rejected_at"
        ] = pd.Timestamp.utcnow()

        df_rejeitados[
            "_rejected_year"
        ] = ano

    return (
        df_validos,
        df_brasil,
        df_rejeitados
    )


metas_ufs_validas = {}
metas_ufs_brasil = {}
metas_ufs_rejeitadas = {}

for ano, df in [
    (2023, df_metas_ufs_2023),
    (2024, df_metas_ufs_2024),
    (2025, df_metas_ufs_2025)
]:

    (
        metas_ufs_validas[ano],
        metas_ufs_brasil[ano],
        metas_ufs_rejeitadas[ano]
    ) = separar_metas_ufs(
        df=df,
        ano=ano
    )

    print(
        f"Ano {ano} | "
        f"registros de origem: {len(df)} | "
        f"UFs válidas: {len(metas_ufs_validas[ano])} | "
        f"Brasil: {len(metas_ufs_brasil[ano])} | "
        f"rejeitados: {len(metas_ufs_rejeitadas[ano])}"
    )
````

## 1.8 Persistência do agregado Brasil e dos rejeitados

O agregado Brasil é preservado como produto separado, pois possui granularidade nacional.

Os registros sem código de UF e sem identificação nacional são armazenados na área de rejeitados para auditoria.

````python
# Objetivo:
# Persistir o agregado Brasil e os
# registros rejeitados da entidade
# Metas por UF.

# Justificativa:
# O agregado nacional não deve compor
# a base estadual, mas precisa ser
# preservado. Registros inválidos
# também devem permanecer auditáveis.

# Ação:
# Grava os agregados e rejeitados por
# ano em diretórios específicos.

CAMINHO_AGREGADO_BRASIL = (
    CAMINHO_SILVER
    / "agregado_brasil"
)

CAMINHO_REJEITADOS_UFS = (
    LOG_PATH
    / "rejected"
    / "silver"
    / "metas_ufs"
)

for ano in [2023, 2024, 2025]:

    destino_brasil = (
        CAMINHO_AGREGADO_BRASIL
        / f"ano={ano}"
    )

    destino_brasil.mkdir(
        parents=True,
        exist_ok=True
    )

    nome_brasil = (
        f"TS_METAS_BRASIL_"
        f"{ano}_SILVER.csv"
    )

    metas_ufs_brasil[ano].to_csv(
        destino_brasil
        / nome_brasil,
        sep=";",
        decimal=",",
        encoding="utf-8",
        index=False
    )

    destino_rejeitados = (
        CAMINHO_REJEITADOS_UFS
        / f"ano={ano}"
    )

    destino_rejeitados.mkdir(
        parents=True,
        exist_ok=True
    )

    nome_rejeitados = (
        f"TS_METAS_UFS_{ano}_"
        f"REJEITADOS.csv"
    )

    metas_ufs_rejeitadas[ano].to_csv(
        destino_rejeitados
        / nome_rejeitados,
        sep=";",
        decimal=",",
        encoding="utf-8",
        index=False
    )

    print(
        f"Ano {ano} | "
        f"Brasil: {len(metas_ufs_brasil[ano])} | "
        f"rejeitados: {len(metas_ufs_rejeitadas[ano])}"
    )
````

````python
# Objetivo:
# Persistir os registros estaduais
# válidos da entidade metas_ufs na
# camada Silver.

# Justificativa:
# A persistência utiliza os caminhos
# e nomes registrados na
# silver_metadata. O agregado Brasil
# e os registros inválidos permanecem
# armazenados separadamente.

# Ação:
# Grava somente os registros com
# código de UF preenchido.

for ano, df in (
    metas_ufs_validas.items()
):

    registro = metadata_dataset[
        metadata_dataset["ano"] == ano
    ].iloc[0]

    destino = Path(
        registro["silver_path"]
    )

    nome_arquivo = registro[
        "silver_file_name"
    ]

    destino.mkdir(
        parents=True,
        exist_ok=True
    )

    df.to_csv(
        destino / nome_arquivo,
        sep=";",
        decimal=",",
        encoding="utf-8",
        index=False
    )

    print(
        f"Silver Metas UFs {ano}: "
        f"{len(df)} registros estaduais válidos | "
        f"{destino / nome_arquivo}"
    )
````
